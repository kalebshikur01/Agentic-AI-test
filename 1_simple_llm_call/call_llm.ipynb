```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
response = llm.invoke("How many moons does jupiter planet have?")
print(response)
```

```python
response = llm.invoke([
"system","You are a helpful Brief assistance who answer questions with one sentence",
"user","How many moons does jupiter planet have?"
])

print(response)
```

```python
messages = [
    {"role":"system","content":"You are a helpful Brief assistance who answer questions with one sentence"},
    {"role":"user","content":"How many moons does jupiter planet have?"}
]
response = llm.invoke(messages)
print(response)
```
