```python
from dotenv import load_dotenv
```

```python
from langchain_openai import ChatOpenAI
load_dotenv()
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
with open("healthcare.txt", "r") as file:
    report=file.read()

print(report[0:250])
```

```python
extraction_prompt = f"""
You are a medical data extraction assistant.

from laboratory report, extract all test values and classify each one as high, low, or normal
based on the reference ranges provided in the report.

Format your response as:
- Test Name: value | Status: HiGH/LOW/NORMAL | Reference: range


Blood Report:
{report}
"""


extraction_response = llm.invoke(extraction_prompt)
extraction_values = extraction_response.text

print("=== STAGE 1: EXTRACTED VALUES ===")
print(extraction_values)
```

```python
diet_prompt = f"""
You are a clinical nutritionist specializing in Ethiopian dietry habits.

Base on the Lab result, write:
1. A short health summary in 4-5 lines explaining the patient's condition in simple languate
2. A short, practical Ethiopian diet plan having only two sections (1) Food to eat (2) Foods to avoid.
Do not include any other sections in diet plan.

Blood Work Analysis:
{extraction_values}
"""

diet_response = llm.invoke(diet_prompt)

print("=== STAGE 2: HEALTH SUMMARY & DIET PLAN ===")
print(diet_response.text)
```
